# **🔍 EDA — Question 1: Overall Sales & Profit**

```python
import pandas as pd
import numpy as np
```

# **Load Dataset**

```python
from google.colab import drive

drive.mount('/content/drive')
```

```python
file_path = "/content/drive/MyDrive/Sales_Performance_Analytics/data/Superstore.csv"

df = pd.read_csv(file_path, encoding="latin1")

df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date'] = pd.to_datetime(df['Ship Date'])

df['Shipping Days'] = (
    df['Ship Date'] - df['Order Date']
).dt.days

df['Order Year'] = df['Order Date'].dt.year
df['Order Month'] = df['Order Date'].dt.month
df['Order Month Name'] = df['Order Date'].dt.month_name()
```

# **Overall Sales**

```python
total_sales = df['Sales'].sum()

total_sales
```

# **Overall Profit**

```python
total_profit = df['Profit'].sum()

total_profit
```

# **Average Sales & Profit**

```python
average_sales = df['Sales'].mean()
average_profit = df['Profit'].mean()

print("Average Sales:", average_sales)
print("Average Profit:", average_profit)
```

# **Create Summary**

```python
summary = pd.DataFrame({
    'Metric': [
        'Total Sales',
        'Total Profit',
        'Average Sales',
        'Average Profit'
    ],
    'Value': [
        total_sales,
        total_profit,
        average_sales,
        average_profit
    ]
})

summary
```

# **🔹 Q2 — Category-wise Sales & Profit**

**Category-wise Sales**

```python
category_sales = df.groupby('Category')['Sales'].sum()

category_sales
```

**Category-wise Profit**

```python
category_profit = df.groupby('Category')['Profit'].sum()

category_profit
```

**Summary**

```python
category_summary = df.groupby('Category').agg(
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum')
)

category_summary
```

Category

   ↓

Total Sales

   ↓
   
Total Profit

## Category-wise Sales & Profit Insights

- Technology generated the highest total sales and total profit among all categories.
- Office Supplies generated lower sales than Technology but achieved a strong profit.
- Furniture generated substantial sales but comparatively lower profit.
- The difference between sales and profit across categories requires further analysis.

# **🔹 Q3 — Region-wise Sales & Profit**

```python
region_summary = df.groupby('Region').agg(
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum')
)

region_summary
```

## Region-wise Sales & Profit Insights

- The West region generated the highest total sales and total profit.
- The East region recorded the second-highest sales and profit.
- The South region generated the lowest total sales among all regions.
- The Central region generated higher sales than the South region but comparatively lower profit.

# **🔹 Q4 — Top 10 Products by Sales**

```python
top_products = (
    df.groupby('Product Name')['Sales']
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

top_products
```

## Top 10 Products by Sales — Insights

- Canon imageCLASS 2200 Advanced Copier generated the highest total sales among all products.
- The top-selling products are largely related to office equipment, printers, copiers, and binding systems.
- The Canon imageCLASS 2200 Advanced Copier shows significantly higher sales than the other top-performing products.

# **🔹 Q5 — Top 10 Customers by Sales**

```python
top_customers = (
    df.groupby('Customer Name')['Sales']
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

top_customers
```

## Top 10 Customers by Sales — Insights

- Sean Miller generated the highest total sales among all customers.
- Tamara Chand ranked second in total sales.
- The top customers represent important contributors to overall sales.
- Customer-level sales analysis can help identify high-value customers.

# **🔹 Q6 — Year-wise Sales**

```python
yearly_sales = (
    df.groupby('Order Year')['Sales']
      .sum()
)

yearly_sales
```

```python
yearly_sales.sort_values(ascending=False)
```

## Year-wise Sales Insights

- 2017 recorded the highest total sales of 733,215.26.
- 2015 recorded the lowest total sales of 470,532.51.
- Sales increased significantly from 2015 to 2016 and continued to grow in 2017.
- The overall trend indicates strong sales growth toward the later years of the dataset.

# **🔹 Q7 — Year-wise Profit**

```python
yearly_profit = (
    df.groupby('Order Year')['Profit']
      .sum()
)

yearly_profit
```

## Year-wise Profit Insights

- 2017 recorded the highest total profit of 93,439.27.
- 2014 recorded the lowest total profit of 49,543.97.
- Total profit increased consistently from 2014 to 2017.
- The growth in profit indicates improving overall profitability over the analyzed period.

# **🔹 Q8 — Monthly Sales Trend**

```python
monthly_sales = (
    df.groupby('Order Month')['Sales']
      .sum()
)

monthly_sales
```

```python
monthly_sales_named = (
    df.groupby(['Order Month', 'Order Month Name'])['Sales']
      .sum()
      .sort_index()
)

monthly_sales_named
```

## Monthly Sales Trend — Insights

- November recorded the highest monthly sales of 352,461.07.
- February recorded the lowest monthly sales of 59,751.25.
- September, November, and December recorded strong sales performance.
- Sales were relatively higher during the later months of the year.
- The monthly analysis shows noticeable variation in sales across different months.

# **Q9 — Total Quantity Sold by Category.**

```python
category_quantity = (
    df.groupby('Category')['Quantity']
      .sum()
)

category_quantity
```

```python
category_quantity.sort_values(ascending=False)
```

## Category-wise Quantity Sold — Insights

- Office Supplies recorded the highest quantity sold with 22,906 units.
- Furniture recorded 8,028 units sold.
- Technology recorded 6,939 units sold.
- Although Office Supplies had the highest quantity sold, Technology generated the highest sales and profit.
- This indicates that sales performance depends not only on quantity sold but also on the value of individual products.

# **🔹 Q10 — Top 10 Products by Profit**
